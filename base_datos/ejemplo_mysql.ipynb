```python
import mysql.connector  # Importar el conector de MySQL
from mysql.connector import Error
```

Para probar que la conexión funciona

```python
# Abrir una conexión a la base de datos MySQL
try:
    connection = mysql.connector.connect(
        host='localhost', # Cambia esto si tu máquina es diferente. Localhost suele funcionar si MySQL está en tu máquina.
        database='eoi', # Asegúrate de que esta base de datos exista en tu servidor MySQL
        user='root', # Cambia esto si tu usuario es diferente. A menudo es 'root' por defecto en una instalación de Xampp
        password='' # Cambia esto si tu contraseña es diferente. A menudo es una cadena vacía por defecto en una instalación de Xampp
    )

    if connection.is_connected():
        db_info = connection.get_server_info()
        print("Conexión exitosa a la base de datos MySQL. Versión del servidor:", db_info)

        # Crear un cursor para ejecutar consultas
        cursor = connection.cursor()
        cursor.execute("SELECT DATABASE();")
        record = cursor.fetchone()
        print("Conectado a la base de datos:", record)

except Error as e:
    print("Error al conectar a MySQL:", e)

       
```

En la siguiente celda, ejecutaremos la Select que consideremos oportuna sobre la base de datos

```python
try:
    if connection.is_connected():
        cursor = connection.cursor(dictionary=True)
        sql = "Select * from clientes"
        cursor.execute(sql)
        resultados = cursor.fetchall()
        for fila in resultados:
            print(f"Empresa: {fila['empresa']}, Contacto: {fila['limite_credito']}")
except Error as e:
    print("Error al ejecutar la consulta:", e)
```

Cerramos la conexión con la base de datos

```python
if connection.is_connected():
    cursor.close()
    connection.close()
    print("Conexión a MySQL cerrada.")
```
