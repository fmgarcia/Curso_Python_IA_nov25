```python
import pandas as pd
from sqlalchemy import create_engine
```

```python

host='localhost' # Cambia esto si tu máquina es diferente. Localhost suele funcionar si MySQL está en tu máquina.
database='eoi' # Asegúrate de que esta base de datos exista en tu servidor MySQL
user='root' # Cambia esto si tu usuario es diferente. A menudo es 'root' por defecto en una instalación de Xampp
password='' # Cambia esto si tu contraseña es diferente. A menudo es una cadena vacía por defecto en una instalación de Xampp
port=3306  # Puerto por defecto de MySQL
```

Crear la cadena de conexión con la base de datos (URL de conexión)

```python
engine = create_engine(f'mysql+pymysql://{user}:{password}@{host}:{port}/{database}')
try:
    sql = "Select * from clientes"
    df = pd.read_sql(sql, con=engine)
except Exception as e:
    print("Error al ejecutar la consulta:", e)
```
