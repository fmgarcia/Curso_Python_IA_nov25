```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt # Librería para visualización de datos
```

```python
# Datos
meses = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio']
ventas = [2500, 3000, 2800, 3500, 4000, 4200]
gastos = [1500, 1600, 1700, 1800, 2000, 2100]
beneficios = np.array(ventas) - np.array(gastos)
productos = ['Producto A', 'Producto B', 'Producto C', 'Producto D', 'Producto E', 'Producto F']
ventas_productos = [500, 700, 600, 800, 900, 1000]
df = pd.DataFrame({
    'Mes': meses,
    'Ventas': ventas,
    'Gastos': gastos,
    'Beneficios': beneficios,
    'Producto': productos,
    'Ventas_Producto': ventas_productos
})  
```

```python
# Crear la figura del dashboard
fig = plt.figure(figsize=(14,10)) # Crear una figura de 14x10 pulgadas
fig.suptitle('Dashboard de Ventas', fontsize=16, fontweight='bold') # Título principal del dashboard
```

```python
# Gráfico de líneas - Ventas y Gastos Mensuales
ax1 = fig.add_subplot(2, 2, 1) # Subgráfico 1
ax1.plot(df['Mes'], df['Ventas'], marker='o', label='Ventas', color='blue')
ax1.plot(df['Mes'], df['Gastos'], marker='o', label='Gastos', color='red')
ax1.set_title('Ventas y Gastos Mensuales')
ax1.set_xlabel('Mes')
ax1.set_ylabel('Cantidad ($)')
ax1.legend()
ax1.grid(True)
# Gráfico de barras - Beneficios Mensuales
ax2 = fig.add_subplot(2, 2, 2) # Subgráfico 2
ax2.bar(df['Mes'], df['Beneficios'], color='green')
ax2.set_title('Beneficios Mensuales')
ax2.set_xlabel('Mes')
ax2.set_ylabel('Cantidad ($)')
# Gráfico de pastel - Ventas por Producto
ax3 = fig.add_subplot(2, 2, 3) # Subgráfico 3
ax3.pie(df['Ventas_Producto'], labels=df['Producto'], autopct='%1.1f%%', startangle=140)
ax3.set_title('Ventas por Producto')
# Histograma - Distribución de Ventas
ax4 = fig.add_subplot(2, 2, 4) # Subgráfico 4
ax4.hist(df['Ventas'], bins=5, color='purple', edgecolor='black')
ax4.set_title('Distribución de Ventas')
ax4.set_xlabel('Cantidad de Ventas ($)')
ax4.set_ylabel('Frecuencia')
```

```python
# Mostrar el dashboard
plt.tight_layout() # Ajustar el diseño para que no se solapen los elementos
plt.savefig('dashboard_ventas.png') # Guardar la figura como archivo PNG
plt.show() # Mostrar la figura completa
```
