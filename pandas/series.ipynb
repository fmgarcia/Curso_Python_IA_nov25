## Series de Pandas

Las series en pandas son estructuras de datos unidimensionales que pueden contener datos de diferentes tipos (enteros, cadenas, números de punto flotante, objetos de Python, etc.). Cada elemento en una serie tiene un índice asociado que permite acceder a los datos de manera eficiente.

```python
# Crear series
s1 = pd.Series([10, 20, 30, 40, 50]) # Crear serie a partir de una lista directamente
print("Serie s1:")
print(s1)
lista = [5, 15, 25, 35, 45]
s2 = pd.Series(lista) # Crear serie a partir de una lista
print("\nSerie s2:")
print(s2)
diccionario = {'a': 100, 'b': 200, 'c': 300} # Crear serie a partir de un diccionario
s3 = pd.Series(diccionario)
print("\nSerie s3:")
print(s3)
valores = [1, 2, 3, 4, 5]
indices = ['x', 'y', 'z', 'w', 'v']
s4 = pd.Series(valores, index=indices) # Crear serie con índices personalizados
print("\nSerie s4:")
print(s4)
array = np.array([7, 14, 21, 28, 35])
s5 = pd.Series(array) # Crear serie a partir de un array de NumPy
print("\nSerie s5:")
print(s5)
s6 = pd.Series(data=[1,2,3,4,5], index=['a', 'b', 'c', 'd', 'e'], name="Serie6") # Crear serie con datos aleatorios
print("\nSerie s6:")
print(s6)
```

```python
print(s6.values)  # Acceder a los valores de la serie
print(s6.index)   # Acceder a los índices de la serie
print(s6.name)    # Acceder al nombre de la serie (si se ha asignado uno)
print(s6.dtype)   # Acceder al tipo de datos de la serie
print(s6.shape)   # Acceder a la forma (número de elementos) de la serie
print(s6.size)    # Acceder al tamaño (número de elementos) de la serie
print(s6.head(3)) # Mostrar los primeros 3 elementos de la serie
print(s6.tail(2)) # Mostrar los últimos 2 elementos de la serie
print(s6.describe()) # Obtener estadísticas descriptivas de la serie
print(s6.sort_values()) # Ordenar la serie por valores
print(s6.sort_index())  # Ordenar la serie por índices
print(s6.isnull())    # Comprobar si hay valores nulos en la serie
print(s6.notnull())   # Comprobar si hay valores no nulos en la serie
print(s6.unique())    # Obtener los valores únicos de la serie
print(s6.value_counts()) # Contar la frecuencia de cada valor en la serie
print(s6.mean())     # Calcular la media de la serie
print(s6.median())   # Calcular la mediana de la serie
print(s6.sum())      # Calcular la suma de los valores de la serie
print(s6.std())      # Calcular la desviación estándar de la serie
```

```python
print(s6['c'])  # Acceder a un elemento por su índice
print(s6[1])    # Acceder a un elemento por su posición
print(s6.iloc[2]) # Acceder a un elemento por su posición usando iloc
print(s6.loc['d']) # Acceder a un elemento por su índice usando loc
```

```python
s6[1:3]  # Acceder a un rango de elementos por posición
```

```python
s6['a':'c']  # Acceder a un rango de elementos por índice
```

```python
# Operaciones con series
s7 = s6 + 10  # Sumar un valor escalar a cada elemento
print("\nSerie s7 (s6 + 10):")
print(s7)
```

```python
s1 = pd.Series(data=[1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'], name="Serie1")
s2 = pd.Series(data=[10, 20, 30, 40, 50], index=['a', 'b', 'e', 'd', 'f'], name="Serie2")
```

Las operaciones que no coincidan los índices pondrá NaN

```python
s1 + s2  # Sumar dos series (alineación por índice)
```

```python
s1 - s2  # Restar dos series (alineación por índice)
```

Métodos útiles de las series de Pandas

```python
print(s1.sum())  # Sumar todos los elementos de la serie s1
print(s2.mean()) # Calcular la media de la serie s2
print(s1.max())  # Obtener el valor máximo de la serie s1
print(s2.min())  # Obtener el valor mínimo de la serie s2
print(s1.std())  # Calcular la desviación estándar de la serie s1

```

Método apply

```python
alturas = pd.Series({'Alice': 165, 'Bob': 175, 'Charlie': 180, 'David': 170})
print("Alturas:")
print(alturas)
```

```python
alturas_metros = alturas.apply(lambda x: x / 100)  # Convertir alturas de cm a metros (devuelve una nueva serie)
print("\nAlturas en metros:")
print(alturas_metros)
```

```python
habitantes_ciudades = pd.Series({'Madrid': 3223000, 'Barcelona': 1620000, 'Valencia': 794000, 'Sevilla': 688000})
ciudades_analizar = ['Madrid', 'Valencia', 'Bilbao']
serie_analizar  = pd.Series(habitantes_ciudades, index=ciudades_analizar)
print("\nHabitantes de ciudades a analizar (con NaN para ciudades no encontradas):")
print(serie_analizar)
```

```python
pd.isnull(serie_analizar)  # Comprobar qué ciudades tienen valores NaN
```

```python
pd.notnull(serie_analizar)  # Comprobar qué ciudades no tienen valores NaN
```

```python
serie_analizar.isnull()  # Comprobar qué ciudades tienen valores NaN
```

```python
serie_analizar.fillna(0, inplace=True)  # Rellenar valores NaN con 0
print("\nSerie después de rellenar NaN con 0:")
print(serie_analizar)
```

```python
habitantes_ciudades = pd.Series({'Madrid': 3223000, 'Barcelona': 1620000, 'Valencia': 794000, 'Sevilla': 688000})
ciudades_analizar = ['Madrid', 'Valencia', 'Bilbao']
serie_analizar  = pd.Series(habitantes_ciudades, index=ciudades_analizar)
serie_analizar.fillna(serie_analizar.mean(), inplace=True)  # Rellenar valores NaN con la media
```

```python
print("\nSerie después de rellenar NaN con la media:")
print(serie_analizar)
```

```python
print(serie_analizar)
```

```python
serie_analizar.name = 'Población'
serie_analizar.index.name = 'Ciudad'
print("\nSerie con nombre y nombre de índice asignados:")
print(serie_analizar)
```

```python
serie_analizar['Madrid'] += 100000  # Como modificar valores en una serie es mutable
serie_analizar.index = ['MAD', 'VAL', 'BIL'] # Modificar los índices de la serie
print("\nSerie después de modificar valores e índices:")
print(serie_analizar)
```
