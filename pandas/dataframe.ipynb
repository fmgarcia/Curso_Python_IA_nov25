#### Slicing condicional

```python
df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Marta', 'Lucia'],
    'Edad': [28, 34, 29, 42, 23],
    'Sexo': ['F', 'M', 'M', 'F', 'F'],
    'Altura': [1.65, 1.80, 1.75, 1.60, 1.70],
    'Peso': [60, 80, 75, 55, 65]
})
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df[df['Edad'] > 30]  # Filtrar filas donde la edad es mayor a 30
```

```python
df[df.Sexo == 'F']  # Filtrar filas donde el sexo es femenino
```

```python
df[-(df.Sexo == 'F')] # Filtrar filas donde el sexo no es femenino
```

```python
df[(df.Sexo == 'M') & (df['Altura'] > 1.75)]  # Filtrar filas donde el sexo es masculino y la altura es mayor a 1.75
```

```python
df[(df.Altura > 1.70) & (df.Altura < 1.80)]  # Filtrar filas donde la altura está entre 1.70 y 1.80. Condiciones de rango
```

```python
df[(df.Altura > 1.70) | (df.Sexo == 'F')] # Filtrar filas donde la altura es mayor a 1.70 o el sexo es femenino. Condiciones "o"
```

Añadir columna al Dataframe

```python
df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Marta', 'Lucia'],
    'Edad': [28, 34, 29, 42, 23],
    'Sexo': ['F', 'M', 'M', 'F', 'F'],
    'Altura': [1.65, 1.80, 1.75, 1.60, 1.70],
    'Peso': [60, 80, 75, 55, 65]
})
```

```python
df.head()
```

```python
df['IMC'] = df['Peso'] / (df['Altura'] ** 2)  # Calcular el Índice de Masa Corporal (IMC) y agregarlo como una nueva columna
df.head()
```

Merge

```python
df_2_primeras = df[:2] # Seleccionar las dos primeras filas del DataFrame
df_2_primeras
```

```python
df_2_ultimas = df[-2:] # Seleccionar las dos últimas filas del DataFrame
df_2_ultimas
```

```python
df_2_ultimas.loc[3, 'Nombre'] = 'Ana'  # Cambiar el nombre en la fila con índice 3 a 'Ana'
```

```python
df_2_ultimas
```

```python
union = pd.merge(df_2_primeras, df_2_ultimas, left_on='Nombre', right_on='Nombre', how='inner')  # unir los dos DataFrames seleccionados usando una unión interna
union.head()
```

Concatenar por columnas

```python
concatenado = pd.concat([df_2_primeras, df_2_ultimas], ignore_index=True)  # Concatenar los dos DataFrames seleccionados ignorando los índices originales
concatenado
```
