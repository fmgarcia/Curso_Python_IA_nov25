## Operaciones con Arrays en Numpy

```python
import numpy as np
```

```python
a = np.array([1,2,3])
a*3
```

```python
a+a
```

```python
np.repeat(a,3)
```

```python
a = np.arange(10) # 0..9
raiz_a = np.sqrt(a)
raiz_a
```

```python
exp_a = np.exp(a)
exp_a
```

```python
np.random.seed(123) # fijar la semilla
a = np.random.randint(1,1000,100)
a
```

```python
np.unique(a)
```

```python
b = np.array([1,1,2,3,5,4])
print(b.mean())
print(np.mean(b))
print(np.median(b))
print(np.sum(b))
print(np.round(np.var(b),2))
print(np.max(b))
print(np.min(b))
print(np.argmax(b))
print(b.std())
```

```python
y = np.arange(1,10) # 1..9
y
```

```python
y = y.reshape(3,-1) # 3 filas, columnas automáticas
y
```

```python
print(y.mean(axis=0)) # media por columnas
print(y.mean(axis=1)) # media por filas
print(y.max(axis=0)) # max por columnas
print(y.max(axis=1)) # max por filas
```

```python
(y>5).sum()  # cuántos elementos son mayores que 5
```

```python
desordenar = np.array([3,1,4,2,5,1])
np.sort(desordenar)
```

```python
sorted_desordenar = sorted(set(desordenar)) # lista ordenada sin duplicados de un array
sorted_desordenar
```

```python
numeros_desordenados = np.array([10,2,8,4,6,2,4,8,10])
test_a = numeros_desordenados > 3
test_a
```
