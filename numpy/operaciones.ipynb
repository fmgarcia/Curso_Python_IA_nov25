## Operaciones con Arrays en Numpy

```python
import numpy as np
```

```python
a = np.array([1,2,3])
a*3
```

```python
a+a
```

```python
np.repeat(a,3)
```

```python
a = np.arange(10) # 0..9
raiz_a = np.sqrt(a)
raiz_a
```

```python
exp_a = np.exp(a)
exp_a
```

```python
np.random.seed(123) # fijar la semilla
a = np.random.randint(1,1000,100)
a
```

```python
np.unique(a)
```

```python
b = np.array([1,1,2,3,5,4])
print(b.mean())
print(np.mean(b))
print(np.median(b))
print(np.sum(b))
print(np.round(np.var(b),2))
print(np.max(b))
print(np.min(b))
print(np.argmax(b))
print(b.std())
```
