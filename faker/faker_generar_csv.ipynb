```python
from faker import Faker
import csv


fake = Faker('es_ES')  # datos en español ficticios

with open('../ficheros/datos_ficticios.csv', mode='w', newline='', encoding='utf-8') as file:
    fieldnames = ['nombre', 'direccion', 'correo_electronico', 'telefono']
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()

    for _ in range(1000):
        writer.writerow({
            'nombre': fake.name(),
            'direccion': fake.address().replace('\n', ', '),
            'correo_electronico': fake.email(),
            'telefono': fake.phone_number()
        })
    
```
