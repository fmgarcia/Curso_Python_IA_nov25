```python
from faker import Faker
import dumper

fake = Faker() # datos en inglés por defecto ficticios
perfil1 = fake.simple_profile()
dumper.dump(perfil1)

print("-------------------")

perfil2 = fake.simple_profile()
dumper.dump(perfil2)
```

```python
# Faker en español
fake_es = Faker('es_ES') # datos en español ficticios
perfil_es = fake_es.simple_profile()
dumper.dump(perfil_es)
```

```python
fake_es_jp = Faker(['es_ES', 'ja_JP']) # datos en español y japonés ficticios
perfil_es_jp = fake_es_jp.simple_profile()
dumper.dump(perfil_es_jp)
```

```python
# datos ficticios personalizados
fake = Faker()
print(f'Seis palabras aleatorias: {fake.words(nb=6)}')
colores = ['rojo', 'verde', 'azul', 'amarillo', 'negro', 'blanco','morado']
print(f'Tres colores aleatorios: {fake.words(nb=3,ext_word_list=colores,unique=True)}')
print(f'Tres colores aleatorios: {fake.random_elements(elements=colores, length=3, unique=True)}')
```

```python
# identificadores
print(f'Identificador aleatorio uuid4: {fake.uuid4()}')
print(f'identificador aleatorio md5: {fake.md5()}')
print(f'identificador aleatorio sha1: {fake.sha1()}')
print(f'identificador aleatorio sha256: {fake.sha256()}')
```
