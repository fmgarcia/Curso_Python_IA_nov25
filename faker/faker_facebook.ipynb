```python
from faker import Faker
import csv

fake = Faker(['en_US', 'es_ES', 'ja_JP', 'de_DE', 'fr_FR'])
cantidad_usuarios = 1000
ruta_fichero = '../ficheros/usuarios_falsos.csv'
campos = ["id", "nombre", "apellidos", "correo_electronico","password",  "dia","mes","ano", "genero"]
with open(ruta_fichero, mode='w', newline='', encoding='utf-8') as fichero_csv:
    escritor_csv = csv.DictWriter(fichero_csv, fieldnames=campos)
    escritor_csv.writeheader()
    for usuario_id in range(1, cantidad_usuarios + 1):
        perfil = fake.simple_profile()
        escritor_csv.writerow({
            "id": usuario_id,
            "nombre": perfil['name'].split()[0],
            "apellidos": ' '.join(perfil['name'].split()[1:]),
            "correo_electronico": perfil['mail'],
            "password": perfil['username'],
            "dia": perfil['birthdate'].day,
            "mes": perfil['birthdate'].month,
            "ano": perfil['birthdate'].year,
            "genero": perfil['sex']
        })
print(f'Se han generado {cantidad_usuarios} usuarios falsos en el fichero {ruta_fichero}.')
```
