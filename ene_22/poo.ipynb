```python
# Creación de una clase simple
class Cuadrado:
    # Constructor con parámetro
    def __init__(self, lado) -> None:
        self.lado = lado # Inicializamos el atributo lado
    
    # Métodos propios de la clase
    # Cálculo del área
    def area(self) -> float:
        return self.lado ** 2
    # Cálculo del perímetro    
    def perimetro(self) -> float:
        return 4 * self.lado
    
    # Representación en forma de cadena de un objeto de la clase    
    def __str__(self) -> str:
        return f"Cuadrado (lado {self.lado}) y área {self.area()} y perímetro {self.perimetro()}"
    
    # Comparación de dos objetos de la clase
    def __eq__(self, otro) -> bool: 
        return self.lado == otro.lado   # Consideramos que dos cuadrados son iguales si tienen el mismo lado
    # Suma de dos objetos de la clase
    def __add__(self, otro):
        return Cuadrado(self.lado + otro.lado)  # La suma de dos cuadrados da un nuevo cuadrado con lado igual a la suma de los lados
    def __mul__(self, otro):
        return self.area() + otro.area() # La multiplicación de dos cuadrados da un número igual a la suma de las áreas
    def __len__(self):
        return len(str(self.lado)) # No tiene mucho sentido calcular la longitud de un cuadrado, pero devuelve la longitud del lado como cadena
        
Cuadrado1 = Cuadrado(10)
Cuadrado2 = Cuadrado(10)
Cuadrado3 = Cuadrado2
print(f"Lado del cuadrado: {Cuadrado1.lado} y su área: {Cuadrado1.area()} y su perímetro: {Cuadrado1.perimetro()}")
print(f"Lado del cuadrado: {Cuadrado2.lado} y su área: {Cuadrado2.area()} y su perímetro: {Cuadrado2.perimetro()}")
print(Cuadrado1)
if(Cuadrado3==Cuadrado2):
    print("Son iguales")
else:
    print("No son iguales")
if(Cuadrado1==Cuadrado2):
    print("Son iguales")
else:
    print("No son iguales")
    
# Sumar dos cuadrados
print(Cuadrado1 + Cuadrado2)
Cuadrado_multiplicado = Cuadrado1 * Cuadrado2
print(Cuadrado_multiplicado) # Nos da la suma de las áreas, 200
print(len(Cuadrado1)) # Nos da un mensaje de error personalizado
```

```python
# Atributos protegidos (convención). Lenguaje de comunicación entre programadores Mangling.
class Cuadrado:
    # Ejemplo atributo protegido con '_'
    def __init__(self, lado) -> None:
        self._lado = lado

c1 = Cuadrado(4)
print(c1._lado) # 4
c1._lado = -1
print(c1._lado) # -1
```

```python
class Cuadrado:
    def __init__(self, lado) -> None:
        self.__lado = lado
        self.area = lado * lado
        self.perimetro = 4 * lado
    def get_lado(self):
        return self.__lado
    def set_lado(self, lado):
        if lado < 1:
            raise ValueError("El lado no puede ser menor a 1")
        self.__lado = lado
        self.area = lado * lado
        self.perimetro = 4 * lado
 
try:       
    c1 = Cuadrado(4)
    print("El valor del lado es: ", c1.get_lado(), " y del área es: ", c1.area, " y del perímetro es: ", c1.perimetro) 
    c1.set_lado(8)
    print("El valor del lado es: ", c1.get_lado(), " y del área es: ", c1.area, " y del perímetro es: ", c1.perimetro) 
    print(c1.get_lado()) # 8
    c1.set_lado(-5) # ValueError: El lado no puede ser menor a 1
except ValueError as ve:
    print(ve)
```

```python
class Cuadrado:
    def __init__(self, lado) -> None:
        self.__lado = lado
    
    @property  # Getter
    def lado(self):
        return self.__lado
    @lado.setter # Setter
    def lado(self, lado):
        if lado < 1:
            raise ValueError("El lado no puede ser menor a 1")
        self.__lado = lado
    @property
    def area(self) -> int: # Ejemplo de propiedad de solo lectura computada
        return self.lado**2
    @property
    def perimetro(self) -> int: # Ejemplo de propiedad de solo lectura computada
        return 4 * self.lado

try:       
    c1 = Cuadrado(4)
     
    c1.lado = 8
     
    print(c1.lado) # 8
    #c1.lado = -5 # ValueError: El lado no puede ser menor a 1
    print(c1.area) # 64
    print(c1.perimetro) # 32
    # c1.area = 10 # AttributeError: can't set attribute realmente porque no existe como tal, se basa en un cálculo de la propiedad lado
except ValueError as ve:
    print(ve)
```
