### Usar expresiones regulares para búsquedas en cadenas con lectura de ficheros

```python
import re

minutos_min = 120 # Mínimo de minutos. 2 horas
exp = r'(?P<minutos>\d{3}) (min)'

with open('../ficheros/Marvel.csv', 'r') as f:
    peliculas_hulk = [linea.split(',')[0] + " - " + linea.split(',')[1] + " - " + re.search(exp, linea).group('minutos') + " minutos - " + linea.split(',')[-1] 
                      for linea in f.readlines() 
                      if re.search(exp, linea) != None and int(re.search(exp, linea).group('minutos')) >= minutos_min] # Al menos tiene más de 120 minutos
for pelicula in peliculas_hulk:
    print(pelicula)
```

```python
# Películas con valoración mayor o igual a un valor dado
# Crearemos un diccionario con las películas que cumplan la condición
valoracion_min = 7 # Valoración mínima

with open('../ficheros/Marvel.csv', 'r') as f:
    lineas_fichero = [linea.strip() for linea in f.readlines()]  # Leemos todas las líneas y eliminamos saltos de línea

peliculas_valoradas = [{'Nombre': linea.split(',')[1], 
                            'Valoración': linea.split(',')[-1]} 
                           for linea in lineas_fichero[1:]
                           if float(linea.split(',')[-1]) >= valoracion_min] # Al menos tiene la valoración mínima
peliculas_valoracion_descendente = sorted(peliculas_valoradas, key=lambda x: x['Valoración'], reverse=True)
print(peliculas_valoracion_descendente)
```

## Pathlib

```python
from pathlib import Path
directorio_salida = Path('./salidas') # Directorio de salida
if not directorio_salida.exists():
    directorio_salida.mkdir()
elif not directorio_salida.is_dir():
    print(f"Error: la ruta {directorio_salida} existe pero no es un directorio")
nombre_fichero_salida = directorio_salida.joinpath('marvel_depurado.csv') # ./salidas/archivo.txt
print(nombre_fichero_salida.absolute()) # Ruta absoluta del fichero de salida
with open(nombre_fichero_salida, 'w') as f:
    f.write("Nombre;Valoración\n")
    for pelicula in peliculas_valoracion_descendente:
        f.write(f"{pelicula['Nombre']};{pelicula['Valoración']}\n")
```

## CSV

```python
import csv
valoracion_min = 7 # Valoración mínima
with open('../ficheros/Marvel.csv', 'r') as f:
    lector_csv = csv.DictReader(f, delimiter=',',quotechar='"')
    peliculas_valoradas = [fila for fila in lector_csv if float(fila['IMDB_Score']) >= valoracion_min]
    peliculas_valoracion_descendente = sorted(peliculas_valoradas, key=lambda x: x['IMDB_Score'], reverse=True)
    print(peliculas_valoracion_descendente)
```

```python
# Filtrado avanzado con múltiples condiciones
# Filtros: valoración mínima, tiempo mínimo y género
import re
import csv
valoracion_min = 7 # Valoración mínima
tiempo_min = 120 # Minutos mínimos
genero = 'Action' # Género a filtrar
filtro_tiempo = lambda x: re.search(r'(?P<minutos>\d{3}) (min)', x['RunTime']) != None and int(re.search(r'(?P<minutos>\d{3}) (min)', x['RunTime']).group('minutos')) >= tiempo_min
filtro_genero = lambda x: genero in x['Genre']
filtro_valoracion = lambda x: float(x['IMDB_Score']) >= valoracion_min
# Lectura del fichero y aplicación de los filtros
with open('../ficheros/Marvel.csv', 'r') as f:
    lector_csv = csv.DictReader(f, delimiter=',',quotechar='"')
    peliculas_filtradas = [fila for fila in lector_csv if filtro_valoracion(fila) and filtro_tiempo(fila) and filtro_genero(fila)]
# Guardar las películas filtradas en un nuevo fichero CSV
with open('./salidas/marvel_filtrado.csv', 'w', newline='') as f:
        campos = ['ID', 'Movie', 'Year', 'Genre', 'RunTime', 'Description', 'IMDB_Score']
        escritor_csv = csv.DictWriter(f, fieldnames=campos, delimiter=',', quotechar='"')
        escritor_csv.writeheader()
        for pelicula in peliculas_filtradas:
            escritor_csv.writerow(pelicula)
```
