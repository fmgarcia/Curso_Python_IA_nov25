## Excepciones

```python
lista = [1,4]
try:
    div = lista[1]/lista[0]
    print(div)
except ZeroDivisionError:
    print("División por cero")
except IndexError:
    print("Índice fuera de rango")
```

```python
lista = [0,4]
try:
    div = lista[1]/lista[0]
    print(div)
except (ZeroDivisionError, IndexError) as e:
    print(f"Problemas: {e}")
```

```python
lista = [0,4]
try:
    div = lista[1]/lista[0]
    print(div)
except Exception as e: # Capturamos cualquier excepción
    print(f"Problemas: {e}")
```

```python
lista = [0,4]
try:
    div = lista[1]/lista[0]
    print(div)
except (ZeroDivisionError, IndexError) as e:
    print(f"Excepeciones individuales: {e}")
except Exception as e: # Capturamos cualquier excepción
    print(f"Problemas: {e}")
finally:
    print("Ejecución finalizada")
```

```python
# Propagación de las excepciones
def dividir(x,y): # Método que divide números enteros positivos
    if not isinstance(x,int) or not isinstance(y,int):
        raise TypeError("Ambos parámetros deben ser de tipo entero")
    if y==0:
        raise ValueError("El divisor no puede ser 0")
    if x<0 or y<0:
        raise ArithmeticError("Los números no pueden ser negativos")
    return x/y


try:
    #dividir("12","43") # se provoca excepcion TypeError en divide y la controla el bloque la ha llamado
    dividir(4,-1)
except TypeError as e:
    print(f"Type error: {e}")
except ArithmeticError as e:
    print(f"Error: {e}")
```
