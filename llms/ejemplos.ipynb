### OpenAI

Voy a hacer la prueba con el modelo gpt-4o-mini

```python
from openai import OpenAI
```

```python
def consultar_openai(prompt, model="gpt-4o-mini"):
    try:
        if not OPENAI_API_KEY:
            return "La clave de API de OpenAI no está configurada."
        client = OpenAI(api_key=OPENAI_API_KEY) # Inicializa el cliente de OpenAI con la clave de API
        response = client.chat.completions.create(
            model=model,
            messages=[
                {"role": "system", "content": "Eres un asistente útil y preciso."},
                {"role": "user", "content": prompt}
            ],
            temperature=0.7 # Controla la creatividad de las respuestas (0.0 a 2.0, siendo 0.0 la más precisa y 2.0 la más creativa)
        )
        return response.choices[0].message.content # Devuelve el contenido de la respuesta generada por el modelo
    except Exception as e:
        return f"Error al consultar OpenAI: {str(e)}"
```

Probar la llamada al método

```python
print("---Respuesta de OpenAI---")
respuesta_openai = consultar_openai("Explica brevemente qué es la inteligencia artificial.")
print(respuesta_openai)
```

## GOOGLE

Voy a hacer la prueba con gemini-2.5-flash

```python
import google.generativeai as genai
```

```python
def consultar_google(prompt, model="gemini-2.5-flash"):
    try:
        if not GOOGLE_API_KEY:
            return "La clave de API de Google no está configurada."
        genai.configure(api_key=GOOGLE_API_KEY) # type: ignore # Configura la clave de API para Google Generative AI
        modelo = genai.GenerativeModel(model) # type: ignore # Crea una instancia del modelo generativo de Google
        response = modelo.generate_content(prompt) # Genera contenido basado en el prompt proporcionado
        return response.text # Devuelve el texto generado por el modelo
    except Exception as e:
        return f"Error al consultar Google: {str(e)}"
```

```python
print("---Respuesta de Google---")
respuesta_google = consultar_google("Explica brevemente qué es la inteligencia artificial.")
print(respuesta_google)
```

## Anthropic

Voy a hacer la prueba con claude-opus-4-6

```python
import anthropic
```

```python
def consultar_claude(prompt, model="claude-opus-4-6"):
    try:
        if not ANTHROPIC_API_KEY:
            return "La clave de API de Anthropic no está configurada."
        client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY) # Inicializa el cliente de OpenAI con la clave de API
        message = client.messages.create(
            model=model,
            max_tokens=1000,
            system = "Eres un asistente útil y preciso.",
            messages=[
                {"role": "user", "content": prompt}
            ],
            temperature=0.7 # Controla la creatividad de las respuestas (0.0 a 2.0, siendo 0.0 la más precisa y 2.0 la más creativa)
        )
        return message.content[0].text # type: ignore # Devuelve el contenido de la respuesta generada por el modelo
    except Exception as e:
        return f"Error al consultar Anthropic: {str(e)}"
```

```python
print("---Respuesta de Anthropic---")
respuesta_anthropic = consultar_claude("Explica brevemente qué es la inteligencia artificial.")
print(respuesta_anthropic)
```

```python
modelos_a_probar = [
    {"nombre": "OpenAI GPT-4o Mini", "funcion": consultar_openai, "modelo": "gpt-4o-mini"},
    {"nombre": "Google Gemini 2.5 Flash", "funcion": consultar_google, "modelo": "gemini-2.5-flash"},
    {"nombre": "Anthropic Claude Opus 4-6", "funcion": consultar_claude, "modelo": "claude-opus-4-6"}
]

prompt = "Explica brevemente qué es la inteligencia artificial."

for modelo in modelos_a_probar:
    print(f"---Respuesta de {modelo['nombre']}---")
    respuesta = modelo["funcion"](prompt, model=modelo["modelo"])
    print(respuesta)
    print("\n" + "-"*50 + "\n")
```
