### Herencia

```python
class Persona():
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        return f"Hola, mi nombre es {self.nombre} y tengo {self.edad} años."
    
```

```python
class Alumno(Persona):
    pass
```

```python
class Estudiante(Alumno):
    def __init__(self, nombre, edad, carrera):
        super().__init__(nombre, edad)
        self.carrera = carrera
    def estudiar(self):
        return f"{self.nombre} está estudiando {self.carrera}."
    def asistir_curso(self):
        return f"{self.nombre} está asistiendo a su curso."
```

```python
class Estudiante_EOI(Estudiante):
    def __init__(self, nombre, edad, carrera, id_eoi):
        super().__init__(nombre, edad, carrera)
        self.id_eoi = id_eoi
    def asistir_curso(self):
        return f"{self.nombre} está asistiendo tiene el ID EOI {self.id_eoi}."
```

```python
class Estudiante_CV(Estudiante):
    def __init__(self, nombre, edad, carrera, nia):
        super().__init__(nombre, edad, carrera)
        self.nia = nia
    def asistir_curso(self):
        return f"{self.nombre} está asistiendo tiene el NIA {self.nia}."
```

```python
class Student(Estudiante):
    pass
```

```python
persona1 = Persona("Ana", 30)
print(persona1.saludar())
#persona1.estudiar()  # Esto generará un error porque Persona no tiene el método estudiar
estudiante1 = Estudiante("Luis", 22, "Ingeniería Informática")
print(estudiante1.saludar())
print(estudiante1.estudiar())
estudiante_eoi1 = Estudiante_EOI("Marta", 25, "Administración", "EOI12345")
print(estudiante_eoi1.saludar())
print(estudiante_eoi1.estudiar())
print(estudiante_eoi1.asistir_curso())
estudiante_cv1 = Estudiante_CV("Carlos", 20, "Derecho", "NIA67890")
print(estudiante_cv1.saludar())
print(estudiante_cv1.estudiar())
print(estudiante_cv1.asistir_curso())
student1 = Student("Elena", 23, "Medicina")
print(student1.saludar())
```

### Ejemplo de polimorfismo

```python
# En este punto, demostramos polimorfismo
# Todas las clases derivadas tienen el método asistir_curso
# Podemos tratarlas de manera uniforme
# Crear una lista de estudiantes de diferentes tipos
# y llamar al mismo método en cada uno

lista_estudiantes = [estudiante1, estudiante_eoi1, estudiante_cv1]
for estudiante in lista_estudiantes:
    print(estudiante.asistir_curso()) # Polimorfismo en acción
```

### Sobrescritura de métodos y atributos

```python
class Trabajador(Persona):
    def __init__(self, nombre, edad, puesto, salario):
        super().__init__(nombre, edad)
        self.puesto = puesto
        self.salario = salario
        
    @property
    def salario_mensual(self):
        return self.salario / 12
        
    def imprimir_salario(self):
        return f"{self.nombre} trabaja como {self.puesto} y gana {self.salario} euros al año."
```

```python
class Trabajador_Nocturno(Trabajador):
    def __init__(self, nombre, edad, puesto, salario, precio_hora_nocturna):
        super().__init__(nombre, edad, puesto, salario)
        self.precio_hora_nocturna = precio_hora_nocturna
        
    def imprimir_salario(self, horas_nocturnas):
        salario_total = self.salario + (self.precio_hora_nocturna * horas_nocturnas)
        return super().imprimir_salario() + f" incluyendo {horas_nocturnas} horas nocturnas, el salario total es {salario_total} euros al año."
```

```python
class Trabajador_PagasExtras(Trabajador):
    @property
    def salario_mensual(self):
        return self.salario / 14
```

```python
trabajador1 = Trabajador("Pedro", 40, "Ingeniero", 50000)
print(trabajador1.imprimir_salario())
print(f"Salario mensual: {trabajador1.salario_mensual} euros")

# Ejemplo de la sobrescritura de métodos
trabajador_nocturno1 = Trabajador_Nocturno("Lucía", 35, "Técnico de mantenimiento", 45000, 20)
print(trabajador_nocturno1.imprimir_salario(1000))  # Suponiendo 1000 horas nocturnas trabajadas

# Ejemplo de la sobrescritura de propiedades
trabajador_pagas_extras = Trabajador_PagasExtras("Ana", 30, "Administrativa", 42000)
print(f"Salario mensual: {trabajador_pagas_extras.salario_mensual} euros")
```

### Relaciones entre clases

```python
lista_personas = [persona1, estudiante1, estudiante_eoi1, estudiante_cv1, trabajador1, trabajador_nocturno1, trabajador_pagas_extras]
for individuo in lista_personas:
    if isinstance(individuo, Persona):
        print(individuo.saludar())
    if isinstance(individuo, Estudiante):
        print(individuo.estudiar())
    if isinstance(individuo, Trabajador):
        print(individuo.imprimir_salario())
    
```

### Herencia Múltiple


```python
class A:
    
    def __init__(self, a1, a2):
        self.a1 = a1
        self.a2 = a2
    
    def metodoA(self):
        return "Método A de la clase A"
    
    def metodo(self):
        return "Método de la clase A"
    
class B:
    
    def __init__(self, b1, b2):
        self.b1 = b1
        self.b2 = b2
    
    def metodoB(self):
        return "Método B de la clase B"
    
    def metodo(self):
        return "Método de la clase B"
```

```python
class C(A, B):
    
    def __init__(self, a1, a2, b1, b2, c1):
        A.__init__(self, a1, a2)
        B.__init__(self, b1, b2)
        self.c1 = c1
    
    def metodoC(self):
        return "Método C de la clase C"
    
    
```

```python
ejemplo_c1 = C("valor_a1", "valor_a2", "valor_b1", "valor_b2", "valor_c1")
print(ejemplo_c1.metodoA())  # Llama al método de la clase A
print(ejemplo_c1.metodoB())  # Llama al método de la clase B
print(ejemplo_c1.metodoC())  # Llama al método de la clase C
print(ejemplo_c1.metodo())    # Llama al método según el orden de resolución de métodos (MRO)
print(C.__mro__)  # Muestra el orden de resolución de métodos (MRO) de la clase C
```

### Mixins

```python
import json

class JsonMixin:
    def to_json(self):
        return json.dumps(self, default = self._serialize_object, indent=2)
    
    def _serialize_object(self, obj):
        if isinstance(obj,(str, int, float, bool, type(None))):
            return obj.__repr__()
        elif isinstance(obj, list):
            return [self._serialize_object(item) for item in obj]
        elif isinstance(obj, dict):
            return {key: self._serialize_object(value) for key, value in obj.items()}
        else:
            return obj.__dict__
        
class Direccion(JsonMixin):
    def __init__(self, calle, ciudad, codigo_postal):
        self.calle = calle
        self.ciudad = ciudad
        self.codigo_postal = codigo_postal
        
class Persona(JsonMixin):
    def __init__(self, nombre, edad, direccion: Direccion):
        self.nombre = nombre
        self.edad = edad
        self.direccion = direccion
        
class Empleado():
    def __init__(self, nombre, edad, direccion: Direccion, puesto, salario):
        self.nombre = nombre
        self.edad = edad
        self.puesto = puesto
        self.salario = salario
        self.direccion = direccion
        

```

```python
direccion1 = Direccion("Calle Falsa 123", "Springfield", "12345")
persona1 = Persona("Homero Simpson", 39, direccion1)
print(persona1.to_json())
print(direccion1.to_json())
empleado1 = Empleado("Marge Simpson", 36, direccion1, "Ama de casa", 30000)
#print(empleado1.to_json())
```
