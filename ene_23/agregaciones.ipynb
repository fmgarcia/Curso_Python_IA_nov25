### Agregaciones/Composiciones

```python
class Direccion:
    def __init__(self, calle, numero, ciudad, codigo_postal):
        self.calle = calle
        self.numero = numero
        self.ciudad = ciudad
        self.codigo_postal = codigo_postal

    def __str__(self):
        return f"{self.calle} {self.numero}, {self.ciudad}, {self.codigo_postal}"
```

```python
class Persona:
    def __init__(self, nombre: str, edad: int, direccion: Direccion):
        self.nombre = nombre
        self.edad = edad
        self.direccion = direccion  # Composición: Persona tiene una Dirección

    def __str__(self):
        return f"{self.nombre}, {self.edad} años, vive en {self.direccion}"
```

```python
class Alumno:
    def __init__(self, nia: str, curso: str, persona: Persona):
        self.nia = nia
        self.curso = curso
        self.persona = persona  # Composición: Alumno tiene una Persona
        
    def __str__(self):
        return f"NIA: {self.nia}, {self.persona}, inscrito en {self.curso}"
```

```python
class Clase:
    def __init__(self, nombre: str, aula: str, alumnos: list[Alumno]):
        self.nombre = nombre
        self.aula = aula
        self.alumnos = alumnos  # Composición: Clase tiene una lista de Alumnos

    def __str__(self):
        return f"Clase {self.nombre} en aula {self.aula} con {len(self.alumnos)} alumnos."
```
