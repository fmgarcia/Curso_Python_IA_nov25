### Dataclass

```python
from dataclasses import dataclass

@dataclass
class Direccion: # Definición de la clase Direccion usando dataclass. Añade automáticamente métodos como __init__ y __repr__ (hace lo mismo que str).
    calle: str
    numero: int
    ciudad: str
    codigo_postal: str
    
    def longitud_calle(self) -> int:
        return len(self.calle)
```

```python
direccion_juan = Direccion("Calle Falsa", 123, "Springfield", "12345")
print(direccion_juan)
```
