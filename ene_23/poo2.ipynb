```python
class Alumno:
    def __init__(self, nombre, edad, curso):
        self.nombre = nombre
        self.edad = edad
        self.curso = curso
```

### Métodos de clase

```python
class Clase:
    numero_alumnos = 0  # Atributo de clase para contar el número de alumnos
    def __init__(self, nombre: str, alumnos: list[Alumno]):
        self.nombre = nombre
        self.alumnos = alumnos  # Composición: Clase tiene una lista de Alumnos
        Clase.numero_alumnos += len(alumnos)
        
    def __str__(self):
        return f"Clase {self.nombre} con {Clase.numero_alumnos} alumnos."
    
    def agregar_alumno(self, alumno: Alumno):
        self.alumnos.append(alumno)
        Clase.numero_alumnos += 1
        
    def dar_baja_alumno(self, alumno: Alumno):
        self.alumnos.remove(alumno)
        Clase.numero_alumnos -= 1
    
    @classmethod
    def obtener_numero_alumnos(cls): # Método de clase.
        return cls.numero_alumnos
```

```python
alumno1 = Alumno("Carlos", 20, "Matemáticas")
alumno2 = Alumno("María", 22, "Física")
alumno3 = Alumno("Lucía", 21, "Química")
alumno4 = Alumno("Javier", 23, "Biología")
lista_alumnos = [alumno1, alumno2, alumno3, alumno4]
clase1 = Clase("Ciencia", lista_alumnos)
print(clase1)
clase1.agregar_alumno(Alumno("Sofía", 20, "Astronomía"))
print(clase1)
clase1.dar_baja_alumno(alumno2)
print(clase1)
print(f"Número total de alumnos: {Clase.obtener_numero_alumnos()}")
```

```python
class Empleado:
    
    sueldo_minimo = 1000 # Atributo de clase (común a todas las instancias). Es una constante estática.
    
    def __init__(self, nombre, edad, sueldo):  # Constructor
        self.nombre = nombre
        self.edad = edad
        self.sueldo = sueldo
        
    @classmethod
    def crea_becario(cls, nombre, edad): # Método de clase como fábrica (factory method).
        return cls(nombre, edad, cls.sueldo_minimo)  # Crea un empleado con sueldo mínimo
```

```python
empleado_becario = Empleado.crea_becario("Pedro", 25)
print(f"Empleado becario: {empleado_becario.nombre}, Edad: {empleado_becario.edad}, Sueldo: {empleado_becario.sueldo}")
```

```python
class Cliente_Gimnasio:
        
    def __init__(self, nombre, edad, membresia, acceso_piscina, acceso_spa, acceso_parking, acceso_clases_grupales):
        self.nombre = nombre
        self.edad = edad
        self.membresia = membresia
        self.acceso_piscina = acceso_piscina
        self.acceso_spa = acceso_spa
        self.acceso_parking = acceso_parking
        self.acceso_clases_grupales = acceso_clases_grupales
        
    @classmethod
    def crea_cliente_basico(cls, nombre, edad):
        return cls(nombre, edad, "Básica", False, False, False, False)
    
    @classmethod
    def crea_cliente_premium(cls, nombre, edad):
        return cls(nombre, edad, "Premium", True, True, True, True)
    
    def __str__(self):
        return f"Cliente: {self.nombre}, Edad: {self.edad}, Membresía: {self.membresia}, Piscina: {self.acceso_piscina}, Spa: {self.acceso_spa}, Parking: {self.acceso_parking}, Clases Grupales: {self.acceso_clases_grupales}"
    
    
cliente1 = Cliente_Gimnasio("Laura", 28, "Particular", True, False, True, False)
cliente2 = Cliente_Gimnasio.crea_cliente_premium("Miguel", 35)
cliente3 = Cliente_Gimnasio.crea_cliente_basico("Ana", 22)
print(cliente1)
print(cliente2)
print(cliente3)
    
```

### Métodos estáticos

```python
class Empleado:
    
    sueldo_minimo = 1000 # Atributo de clase (común a todas las instancias). Es una constante estática.
    
    def __init__(self, nombre, edad, sueldo):  # Constructor
        self.nombre = nombre
        self.edad = edad
        self.sueldo = sueldo
    
    @staticmethod
    def calcular_aumento(sueldo_actual, porcentaje):
        return sueldo_actual + (sueldo_actual * porcentaje / 100)
    
    
```

```python
ana = Empleado("Ana", 30, 1500)
nuevo_sueldo = Empleado.calcular_aumento(ana.sueldo, 10)
print(f"Sueldo actual de {ana.nombre}: {ana.sueldo}")
print(f"Sueldo después del aumento: {nuevo_sueldo}")
```
